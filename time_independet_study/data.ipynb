```python
"""
Heatmap: Theta, Phi vs G.
"""
N_PHI = 200      # number of phi points (x-axis)
N_THETA = 200    # number of theta points (y-axis)

DPARAMETER = 0.3 # dIPR directionality parameter (see AAH_model.dIPR)
phi_lin = np.linspace(0, 2 * np.pi, N_PHI)
theta_lin = np.linspace(0, 2 * np.pi, N_THETA)

out_dir = "data/nonzero_theta"

##############################################################################
# Compute the G(phi) and dIPR(phi) for single values
V1 = 0.5
V2 = 0.5
out_filename = f"G_dIPR_ThetaPhi_N{N}_V1{V1}_V2{V2}.npz"


# Grids
phi_lin = np.linspace(0, 2 * np.pi, N_PHI)
theta_lin = np.linspace(0, 2 * np.pi, N_THETA)
G_matrix = np.zeros((N_THETA, N_PHI))
dIPR_matrix = np.zeros((N_THETA, N_PHI))

for row, theta in enumerate(tqdm.tqdm(theta_lin, desc="theta sweep", unit="row")):

    energies, angles = aah_fun.EnergyPhiSpectrum(
        N=N, V1=V1, V2=V2, phi_lin=phi_lin, t=t, theta=theta, save=False,
    )

    G, _ = aah_fun.G_Phi(
        energies=energies, angles=angles,
        level_i=lvl_i, level_j=lvl_j,
        N=N, V1=V1, V2=V2, theta=theta, t=t, save=False,
    )
    G_matrix[row] = G

    dipr, _ = aah_fun.dIPR_Phi(
        N=N, V1=V1, V2=V2, phi_lin=phi_lin, level_i=lvl_j,
        t=t, theta=theta, dparameter=DPARAMETER, save=False,
    )
    dIPR_matrix[row] = dipr

# Save

os.makedirs(out_dir, exist_ok=True)
full_path = os.path.join(out_dir, out_filename)

np.savez(
    full_path,
    G_matrix=G_matrix,
    dIPR_matrix=dIPR_matrix,
    phi_lin=phi_lin,
    theta_lin=theta_lin,
    N=N, V1=V1, V2=V2, t=t,
    level_i=lvl_i, level_j=lvl_j,
    dparameter=DPARAMETER,
)

print(f"Saved data to {full_path}")

#########################################################################################
# Compute the G(phi) and dIPR(phi) for multiple combinations of V1, V2

# V_values = [0.1, 0.5, 1.4]
# param_combos = [(v1, 0.0) for v1 in V_values] + [(v, v) for v in V_values]
#
# os.makedirs(out_dir, exist_ok=True)
#
# for V1, V2 in param_combos:
#     G_matrix, dIPR_matrix = aah_fun.compute_theta_phi_heatmap(N=N, V1=V1, V2=V2, lvl_i=lvl_i, lvl_j=lvl_j)
#
#     out_filename = f"G_dIPR_ThetaPhi_N{N}_V1{V1}_V2{V2}.npz"
#     full_path = os.path.join(out_dir, out_filename)
#
#     np.savez(
#         full_path,
#         G_matrix=G_matrix,
#         dIPR_matrix=dIPR_matrix,
#         phi_lin=phi_lin,
#         theta_lin=theta_lin,
#         N=N, V1=V1, V2=V2, t=t,
#         level_i=lvl_i, level_j=lvl_j,
#         dparameter=DPARAMETER,
#     )
#
#     print(f"Saved data to {full_path}")
```
